```python
x = [5,6,7,8,9]
y = [11,5,8,6,12]

plt.plot(x,y)
```

# add marker

```python
x = [5,6,7,8,9]
y = [11,5,8,6,12]

plt.plot(x,y, marker = 'o')
plt.show
```

```python
x = [5,6,7,8,9]
y = [11,5,8,6,12]
plt.plot(x,y)
plt.plot(y, marker = '.')
plt.plot(x, marker = '>')
```

# bar graph

```python
x = [1,2,3,4]
y = [5,6,7,8]
c = ['r','g','b','y']
plt.bar(x,y , color = c)
plt.show
```

# horizental bar graph

```python
import numpy as np
```

```python
x = np.array(['a','b','c','d'])
y = np.array([5,6,7,8])
c = ['r','g','b','y']
plt.barh(x,y, color = c)
```

# bar graph width and height

```python
x = np.array(['a','b','c','d'])
y = np.array([5,6,7,8])
c = ['r','g','b','y']
plt.barh(x,y, color = c , height = 0.2)

plt.show()
```

```python
x = [1,2,3,4]
y = [5,6,7,8]
c = ['r','g','b','y']
plt.bar(x,y , color = c, width = 0.2)
plt.show
```

# scatter plot

```python
 import matplotlib.pyplot as plt
```

```python
x = [1,2,3,4,5,6,8,2,3,10]
y = [5,6,7,8,11,15,13,14,5,2]
plt.scatter(x,y)
plt.show()
```

# adding lable

```python
x = [1,2,3,4,5,6,8,2,3,10]
y = [5,6,7,8,11,15,13,14,5,2]
plt.xlabel("number")
plt.ylabel("count")
plt.scatter(x,y)
plt.show()
```

```python
x = [1,2,3,4,5,6,8,2,3,10]
y = [5,6,7,8,11,15,13,14,5,2]
plt.xlabel("number")
plt.ylabel("count")
plt.scatter(x,y, color = 'r')
plt.show()
```

# size change

```python
x = [1,2,3,4,5,6,8,2,3,10]
y = [5,6,7,8,11,15,13,14,5,2]
plt.xlabel("number")
plt.ylabel("count")
plt.scatter(x,y, color = 'r', s = 100)
plt.show()
```
