```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
var   =[1,2,3,4,5,6,7]
var_1 = [8,9,6,5,2,8,6]
```

# using matplotlib 

```python
plt.plot(var ,var_1)
plt.show()
```

# using seaborn

# lineplot

```python
sns.lineplot(x = var, y = var_1)
plt.show()
```

```python
var   =[1,2,3,4,5,6,7]
var_1 = [8,9,6,5,2,8,6]

import pandas as pd 

df = pd.DataFrame({"var":var,"var_1":var_1})
sns.lineplot(x = 'var',y = 'var_1',data = df)
plt.show()
```

# data set load usingseaborn 

```python
df_1 = sns.load_dataset("penguins")
```

```python
df_1=df_1.head(20)
```

```python
df_1
```

# catagorical data divided in male and female using "hue"

```python
sns.lineplot(x = "bill_length_mm",y = "flipper_length_mm",data = df_1,hue ="sex", palette="rocket_r")
plt.show()
```

# marker parameter 

```python
sns.lineplot(x = "bill_length_mm",y = "flipper_length_mm",data = df_1,hue ="sex", palette="rocket_r", style = "sex",markers = ["o",">"])
plt.title("penguine")
plt.grid()
plt.show()
```
