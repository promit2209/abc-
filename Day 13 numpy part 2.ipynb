# slicing in array

# slicing in 1d array

```python
import  numpy as np
arr = np.array([1,2,3,4,5,6])
print(arr[0:3])
print(arr[1:5])
```

# stepping 

```python
arr = np.array([1,2,3,4,5,6,7])
print(arr[::2])
print(arr[::3])
```

# slicing in 2d array

```python
arr = np.array([[1,2,3,4,5],[60,70,80,90,100]])
print(arr[1,1:4])
```

# joining the numpy arrays

# joining for 1d array

```python
arr1 = np.array([1,2,3])

arr2 = np.array([4,5,6])

arr = np.concatenate((arr1,arr2))

print(arr)
```

# joining for 2d array

for axis 1

```python
arr1 = np.array([[1,2,3],[4,5,6]])

arr2 = np.array([[10,20,30],[40,50,60]])

arr = np.concatenate((arr1,arr2),axis = 1)
print(arr)
```

forc axis 0

```python
arr1 = np.array([[1,2,3],[4,5,6]])

arr2 = np.array([[10,20,30],[40,50,60]])

arr = np.concatenate((arr1,arr2),axis = 0)
print(arr)
```

# ravel and flatten

converts the multidimentional array in 1d array

```python
arr = np.array ([[[1,2,3,4,5],[10,20,30,40,50],[100,200,300,400,500]]])
print(arr)
print("dimension is",arr.ndim)
n = arr.ravel()
print(n)
print("dimension of new array",n.ndim)
```

```python
arr = np.array ([[[1,2,3,4,5],[10,20,30,40,50],[100,200,300,400,500]]])
print(arr)
print("dimension is",arr.ndim)
n = arr.flatten()
print(n)
print("dimension of new array",n.ndim)
```

```python
c = np.array([[[[12,23,33],[55,66,88],[89,90,50],[89,55,66]]]])
print(c)
print("dimension is",c.ndim)
n = c.ravel()
print(n)
print("dimension of new array",n.ndim)

```

```python
c = np.array([[[[12,23,33],[55,66,88],[89,90,50],[89,55,66]]]])
print(c)
print("dimension is",c.ndim)
n = c.flatten()
print(n)
print("dimension of new array",n.ndim)
```

# unique function

```python
k = np.array([1,2,3,4,56,8,9,20,54,52,54,56,4,140,45,104])
print(k)
x = np.unique(k)
print(x)
```

# for indexing position ( if we want to print index position )

```python
k = np.array([1,2,3,4,56,8,9,20,54,52,54,56,4,140,45,104])
print(k)
x = np.unique(k,return_index=True)
print(x)
```

# for repetation  count

```python
k = np.array([1,2,3,4,56,8,9,20,54,52,54,56,4,140,45,104])
print(k)
x = np.unique(k,return_index=True, return_counts=True)
print(x)
```

# delete

for 1d array

```python
a = np.array([12,13,14,15])
b = np.delete(a,[1])
print(b)
```

for 2d array

```python
a = np.array([[12,13,14,15],[11,16,17,18],[33,55,66,77]])
print(a)
b = np.delete(a,2, axis=0)
print()
print(b)
```

```python
a = np.array([[12,13,14,15],[11,16,17,18],[33,55,66,77]])
print(a)
b = np.delete(a,2, axis=1)
print()
print(b)
```
