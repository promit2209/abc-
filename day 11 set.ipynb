# join 2 sets 

```python
x = {"data","python","code"}
y = {10,20,30,40,50}

z = x.union(y)
print(z)
```

# write a python code to find a max and min value 


```python
setn = {5,10,6,23,3,5,2,45,1}
print("original set element")
print("max value of the set is:")
print(max(setn))
print("min value of the set is:")
print(min(setn))
```
