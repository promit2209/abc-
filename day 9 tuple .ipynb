# creating a tuple

```python
my_tuple=(1,2,3,4,5,6)
print(my_tuple)
print(type(my_tuple))
```

# mixed data type


```python
my_tuple=(1,2.5,3+5j,4,5.5,"hello",6)
print(my_tuple)
print(type(my_tuple))
```

# empty tuple

```python
my_tuple=()
print(my_tuple)
print(type(my_tuple))
```

# nested tuple

```python
my_tuple=(1,2,3,4,5,6,"data",[1,2,3,4],())
print(my_tuple)
print(type(my_tuple))
```

# slicing 

```python
a = (45,65,9,8,6)
print(a[0:0:1])
print (len(a))
```

# reverse in tuple

```python
a = (45,65,9,8,6)
print(a[::-1])
```

```python
T =(1,2,3,4,5,6,7,8)
print(T[::])
print(T[3:])
print(T[:4])
print(T[-2:-5])
```
