# indexing in 2d array 

```python
arr = np.array([[1,2,3],[4,5,6]])
print ("2nd element of the 1st row:",arr[0,1])
print ("1st element of the 2nd row:",arr[1,0])
```

# add the 1st element of row 1 and last element of the 2nd row

```python
arr = np.array([[1,2,3],[4,5,6]])
print ("2nd element of the 1st row:",arr[0,0])
print ("1st element of the 2nd row:",arr[1,-1])
print("sum:",arr[0,0] + arr[1,-1])
```
