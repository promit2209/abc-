```python
import requests 
import pandas as pd
import json
```

```python
covid = requests.get("https://api.covidtracking.com/v1/us/daily.json")
                
covid_1 = covid.json()
list = []

for i in covid_1:
    
    dic = {"date":"","positive":"","negative":"","hospitalize":""}
    
    date = i.get("date")
    positive = i.get("positive")
    negative = i.get("negative")
    hospitalize = i.get("hospitalize")
    
    dic.update({"date":date,"positive":positive,"negative":negative,"hospitalize":hospitalize})
    list.append(dic)
```
