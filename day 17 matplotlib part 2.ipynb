# matplotlib  pie chart 

```python
import matplotlib.pyplot as plt
```

```python
x = [60,80,70,50]
y = ['english','math','science','bengali']
c = ['yellow','magenta','pink','red']
plt.pie(x, labels = y, colors = c)
plt.legend()
plt.show()
```
