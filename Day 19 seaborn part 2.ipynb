# bar graph in seaborn library

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df = sns.load_dataset("penguins")
```

```python
df
```

```python
sns.barplot(x ='island', y ='bill_length_mm', data = df, hue = "sex")
```

```python
sns.barplot(x =df.island, y =df.bill_length_mm,  hue = df.sex)
```

# order

```python
order_1 = [ "Dream" ,"Torgersen", "Biscoe"]
sns.barplot(x='island', y ='bill_length_mm', data = df, hue = 'sex',order = order_1)
```

# hue order 

```python
order_1 = [ "Dream" ,"Torgersen", "Biscoe"]
sns.barplot(x='island', y ='bill_length_mm', data = df, hue = 'sex',order = order_1,hue_order = ["Female","Male"])
```

# colors 

```python
order_2=["Torgersen","Dream","Biscoe"]
sns.barplot(x='island', y ='bill_length_mm',data = df , hue = "sex" ,hue_order=["Female","Male"],order = order_2,color ="red")
```

```python
order_2=["Torgersen","Dream","Biscoe"]
sns.barplot(x='island', y ='bill_length_mm',data = df , hue = "sex" ,hue_order=["Female","Male"],order = order_2, palette = "Accent")
```

# saturation parameter 

```python
order_2=["Torgersen","Dream","Biscoe"]
sns.barplot(x='island', y ='bill_length_mm',data = df , hue = "sex" ,hue_order=["Female","Male"],order = order_2,color ="red",saturation = 0.50)
```

```python
order_2=["Torgersen","Dream","Biscoe"]
sns.barplot(x='island', y ='bill_length_mm',data = df , hue = "sex" ,hue_order=["Female","Male"],order = order_2,palette = "Accent", saturation = .50)
```
